**O que o código deve fazer:**

## 🥈 Questão 2 — O fluxo de caixa da Rota Sul (1,0)

**Mesa:** Finanças · **Conceito:** acumulador + contador + `if`/`elif` dentro do `while`

A **Rota Sul Logística** fechou seis meses. A receita do mês 1 foi **R$ 12.000,00** e cresce **R$ 800,00 por mês**. O custo do mês 1 foi **R$ 9.000,00** e cresce **R$ 500,00 por mês**, mas há dois eventos:

- no **mês 3**, uma manutenção extraordinária de **R$ 1.500,00**;
- no **mês 5**, um crédito tributário de **R$ 800,00**.

**Tarefa:** com um `while` de 6 meses, calcule:

- receita total, custo total e lucro total do semestre;
- margem líquida (`lucro_total / receita_total * 100`);
- quantos meses tiveram lucro **acima de R$ 4.000,00**.

**Saída esperada:** receita total **R$ 84.000,00**, custo total **R$ 62.200,00**, lucro **R$ 21.800,00**, margem **25,95%**, **2 meses** acima de R$ 4.000,00.

**Dica:** a receita do mês `m` é `12000 + 800 * (m - 1)`. O custo segue a mesma lógica, mas com os eventos dos meses 3 e 5.

````python
mes = 0
receita_total = 0.0
custo_total = 0.0
lucro_total = 0.0
meses_acima = 0

while mes < 6:
    mes += 1

    # TODO: calcule a receita e o custo do mês
    receita = 12000 + 800 * (mes - 1)
    custo = 9000 + 500 * (mes - 1)

    # TODO: aplique os eventos dos meses 3 e 5
    if mes == 3:
        custo = 9000 + 500 * (mes - 1) + 1500 
    elif mes == 5:
        custo = 9000 + 500 * (mes - 1) - 800

    lucro = receita - custo

    # TODO: acumule os totais e conte os meses com lucro > 4000
    receita_total += receita
    custo_total += custo
    lucro_total += lucro
    if lucro > 4000:
        meses_acima += 1
    

    print(f"Mês {mes}: receita R$ {receita:.2f} | custo R$ {custo:.2f} | lucro R$ {lucro:.2f}")

# TODO: calcule a margem e imprima o resumo
margem = lucro_total / receita_total * 100
print(f"Receita total: R$ {receita_total:.2f}")
print(f"Custo total: R$ {custo_total:.2f}")
print(f"Lucro total: R$ {lucro_total:.2f}")
print(f"Margem líquida: {margem:.2f}%")
print(f"Meses com lucro acima de R$ 4.000,00: {meses_acima}")
````

### ✍️ Explique (Q2)

Por que `mes` é incrementado logo no início do loop? O que aconteceria com a receita e o custo do primeiro mês se o incremento ficasse no final?

**Sua resposta:**

Ao incrementar o mês no início do loop, o ciclo começa no mês correto, e os eventos dos meses 3 e 5 são aplicados adequadamente. Caso contrário — se o mês for incrementado ao final —, o loop começará em zero e terminará em 6; ou seja, serão contabilizados 7 meses no total, e não 6.

## 🥉 Questão 3 — O estoque de segurança (1,0)

**Mesa:** Estoque · **Conceito:** `while` decrescente + validação com `raise`

A distribuidora **Sertão Bebidas** tem **2.400** unidades de um produto. A demanda média é de **180 unidades por dia**. O estoque de segurança é de **300 unidades** — abaixo disso, o gerente quer ser avisado para repor.

**Tarefa:** calcule quantos dias o estoque leva para chegar ao nível de segurança e quantas unidades foram vendidas nesse período. Use um `while` **decrescente** (`estoque -= venda_dia`). Antes do loop, valide:

- `estoque > 0`;
- `venda_dia > 0`;
- `estoque_seguranca >= 0`.

Se alguma condição falhar, `raise ValueError` com uma mensagem clara.

**Saída esperada:** **12 dias**, estoque final **240**, **2.160 unidades vendidas**. O estoque final fica abaixo da segurança — é exatamente o gatilho do alerta.

**Dica:** na última venda, o estoque pode ser menor que a venda diária. Trate esse caso com `if`/`else` para não vender mais do que existe.

````python
estoque = 2400
venda_dia = 180
estoque_seguranca = 300

# TODO: valide as três condições com raise ValueError
if estoque <= 0:
    raise ValueError("O estoque inicial deve ser positivo.")
if venda_dia <= 0:
    raise ValueError("A venda diária deve ser positiva.")
if estoque_seguranca < 0:
    raise ValueError("O estoque de segurança não pode ser negativo.")

dias = 0
vendidas = 0

# TODO: troque False pela condição do loop decrescente
while estoque >= estoque_seguranca:
    if estoque < venda_dia:
        # TODO: venda o que resta e zere o estoque
        vendidas = estoque
        estoque = 0
    else:
        # TODO: desconte a venda diária
        vendidas += venda_dia

    # TODO: avance o contador e o acumulador de unidades vendidas
    dias += 1
    estoque -= venda_dia

print(f"Dias até o estoque de segurança: {dias}")
print(f"Estoque final: {estoque} unidades")
print(f"Unidades vendidas no período: {vendidas}")
````

### ✍️ Explique (Q3)

1. Por que a condição do loop é `estoque > estoque_seguranca`, e não `estoque >= 0`?
2. O que mudaria no resultado se `venda_dia` fosse maior que o estoque inicial?

**Sua resposta:**

A condição do loop é que o estoque seja maior ou igual ao estoque de segurança, de modo que, quando o estoque atingir o ponto de estoque de segurança, seja enviado um alerta informando que o estoque chegou ao nível mínimo no qual é necessário realizar a reposição.
Se a venda diária for maior que o estoque inicial, será vendido apenas o que houver em estoque, deixando o estoque zerado.



## 🥈 Questão 4 — O plantão que travou (1,0)

**Mesa:** Finanças pessoais · **Conceito:** loop infinito — diagnóstico e correção

Um estagiário escreveu o código abaixo para simular saques de **R$ 200,00** de um saldo de **R$ 1.050,00**:

```python
saldo = 1050.0
saque = 200.0
while saldo > 0:
    print(f"Saque de R$ {saque:.2f} | saldo: R$ {saldo:.2f}")
```

O programa nunca termina.

> ⚠️ **Não execute o código acima no Colab.** Ele é apenas para leitura; se rodar, vai travar a célula. A correção vai na célula abaixo.

**Tarefa:**

1. explique, em uma frase, **qual linha falta** e por que a condição nunca fica falsa;
2. escreva a versão corrigida, com contador `saques` e tratamento da **última parcela parcial** (quando o saldo é menor que R$ 200,00);
3. responda: se trocássemos `saldo -= valor` por `saldo += valor`, o loop também seria infinito? Por quê?

**Saída esperada:** **6 saques**; os cinco primeiros de R$ 200,00 e o último de **R$ 50,00**; saldo final **R$ 0,00**.

````python
saldo = 1050.0
saque = 200.0
saques = 0

# TODO: escreva o while corrigido.
# Dica: dentro do loop, decida o valor do saque antes de descontar.
while saldo > 0:
    valor = saldo - saque
    if saldo < saque:
        valor = saldo
        saldo = 0
    else:
        saldo -= saque
    saques += 1
    print(f"Saque {saques}: R$ {valor:.2f} | saldo: R$ {saldo:.2f}")

print(f"Total de saques: {saques}")
````

### ✍️ Explique (Q4)

Qual linha faltava no código original? E por que `saldo += valor` também produziria um loop infinito?

**Sua resposta:**

Faltam as linhas `if` e `else`, o que impede que o loop seja interrompido quando o saldo for menor que o valor do saque e também se o saldo mudar para saldo += valor entra em um loop infinito onde o valor aumenta.

## 🥇 Questão 5 — A mesa de ações (1,0)

**Mesa:** Mercado de capitais · **Conceito:** `break`

Um investidor tem **R$ 10.000,00** em caixa e quer comprar lotes de **100 ações** da **PETR4** enquanto o preço estiver **abaixo ou igual a R$ 38,00**. O preço começa em **R$ 35,00** e sobe **R$ 0,90 por dia**.

**Tarefa:** simule as compras com um `while`. Em cada dia:

- se o preço estiver acima do teto, o loop termina pela própria condição;
- se o caixa não cobrir o lote, use `break` e explique o motivo;
- se comprar, desconte o caixa, acumule ações e custo total.

**Saída esperada:** **2 lotes** comprados, **200 ações**, preço médio **R$ 35,45**, caixa final **R$ 2.910,00**. No **dia 3**, o preço é R$ 36,80, mas o caixa não cobre o lote de R$ 3.680,00 — é o `break`.

**Dica:** o `break` não é a única saída. Se o preço passasse de R$ 38,00 antes de o caixa acabar, o loop terminaria pela condição `preco <= teto`.

````python
caixa = 10000.0
preco = 35.00
teto = 38.00
lote = 100
dias = 0
acoes = 0
custo_total = 0.0

while preco <= teto:
    dias += 1
    custo = preco * lote

    # TODO: se o caixa não cobrir o custo, avise e use break
    if caixa < custo:
        print("caixa insuficiente para a compra de ações.")
        break

    # TODO: execute a compra e atualize caixa, ações e custo_total
    caixa -= custo
    acoes += lote
    custo_total += custo

    print(f"Dia {dias}: comprou {lote} ações a R$ {preco:.2f} | caixa: R$ {caixa:.2f}")

    # TODO: atualize o preço para o próximo dia
    preco = preco + 0.90

# TODO: imprima o resumo (só se houver ações)
if acoes > 0:
    preco_medio = custo_total / acoes
    print(f"Total: {acoes} ações | preço médio: R$ {preco_medio:.2f} | caixa: R$ {caixa:.2f}")
else:
    print("Nenhuma compra foi realizada.")
````

### ✍️ Explique (Q5)

O break desta questão aconteceu por causa do preço ou do caixa? Como você sabe? O que aconteceria se o caixa fosse R$ 20.000,00?

**Sua resposta:**

A interrupção foi ativada porque não há mais dinheiro suficiente em caixa para comprar mais ações; é possível saber disso porque aparece uma mensagem que foi programada para esse caso. se a caixa fosse R$ 20.000,00 o resultado é:  Total: 400 ações | preço médio: R$ 36.35 | caixa: R$ 5460.00

## 🥇 Questão 6 — O CTR das campanhas (1,0)

**Mesa:** Marketing · **Conceito:** `continue`

A agência **Vale Criativo** rodou 8 campanhas. As campanhas **3** e **6** não tiveram impressões registradas (dado ausente) e devem ser **ignoradas**, não contadas como CTR zero. Para as demais:

- impressões = `4000 + 1000 * campanha`
- cliques = `120 + 45 * campanha`

**Tarefa:** calcule o **CTR geral** (`cliques / impressões * 100`) e conte quantas campanhas tiveram CTR **maior ou igual a 3,6%**. Use `continue` para pular as campanhas sem dados.

**Saída esperada:** 6 campanhas válidas, **1.935 cliques**, **51.000 impressões**, CTR geral **3,79%**, **4 campanhas** acima da meta.

> 🚨 **Cuidado:** o contador `campanha` deve ser incrementado **antes** do `continue`; caso contrário, o loop fica preso na campanha sem dados.

**Dica:** o `continue` interrompe apenas a rodada atual. Ele não sai do loop e não imprime “fim” — só pula o resto do bloco.

````python
campanha = 0
total_cliques = 0
total_impressoes = 0
validas = 0
acima_meta = 0
meta_ctr = 3.6

while campanha < 8:
    campanha += 1
    impressoes = 4000 + 1000 * campanha
    cliques = 120 + 45 * campanha

    # TODO: zere as impressões das campanhas 3 e 6
    if campanha == 3 or campanha == 6:
        impressoes = 0

    # TODO: se não houver impressões, avise e use continue
    if impressoes == 0:
        print(f"Campanha {campanha}: sem impressões.")
        continue

    ctr = cliques / impressoes * 100

    # TODO: acumule cliques, impressões e conte as campanhas válidas
    total_cliques += cliques
    total_impressoes += impressoes
    validas += 1

    # TODO: conte as campanhas com CTR >= meta_ctr
    if ctr >= meta_ctr:
        acima_meta += 1

    print(f"Campanha {campanha}: CTR = {ctr:.2f}%")

# TODO: calcule e imprima o CTR geral
ctr_geral = total_cliques / total_impressoes * 100 
print(f"Campanhas válidas: {validas}")
print(f"Total de cliques: {total_cliques}")
print(f"Total de impressões: {total_impressoes}")
print(f"CTR geral: {ctr_geral:.2f}%")
print(f"Campanhas acima da meta: {acima_meta}")
````

### ✍️ Explique (Q6)

Por que o `continue` só funciona porque `campanha += 1` vem antes dele? O que aconteceria se o incremento estivesse depois do `continue`?

O fato de campanha += 1 estar antes do continue é o procedimento correto para que o sistema conte cada campanha e pule apenas aquelas que não possuem dados; se fosse feito da forma incorreta — colocando o continue primeiro —, poderia ocorrer um loop infinito.

**Roteiro de teste 2:** `8000`, `9000`, `fim` → total **R$ 17.000,00**, média **R$ 8.500,00**.

> 💡 Esta é a única questão interativa. Execute a célula, digite um valor por vez e pressione `Enter`. Para testar de novo, execute a célula outra vez.

## 🥇 Questão 8 — Cenários de PIB (1,0)

**Mesa:** Macroeconomia · **Conceito:** loops aninhados

A mesa de conjuntura quer projetar o PIB de um estado por **10 anos** em três cenários:

- **Pessimista:** crescimento de **0,5% ao ano**;
- **Base:** **1,8% ao ano**;
- **Otimista:** **3,2% ao ano**.

O PIB inicial é **100,0** (índice). Para cada cenário, o laço interno deve projetar **10 anos** e imprimir o PIB final e o crescimento acumulado. Ao final, mostre qual cenário termina com o **maior PIB**.

**Saída esperada:**

- Pessimista: **105,11** (+5,11%)
- Base: **119,53** (+19,53%)
- Otimista: **137,02** (+37,02%)
- Maior PIB: **Otimista**.

**Dica:** o laço externo decide **qual cenário**; o interno resolve **aquele cenário até o fim**. `pib_cenario` precisa ser reiniciado a cada cenário.

````python
pib_inicial = 100.0
cenario = 0
maior_pib = 0.0
nome_maior = ""

while cenario < 3:
    cenario += 1

    # TODO: defina nome e taxa de cada cenário com if/elif/else
    if cenario == 1:
        nome = "pessimista"
        taxa = 0.5
    elif cenario == 2:
        nome = "base"
        taxa = 1.8
    else:
        nome = "otimista"
        taxa = 3.2

    pib_cenario = pib_inicial
    ano = 0

    # TODO: laço interno de 10 anos
    while ano < 10:
        pib_cenario *= (1 + taxa / 100)
        ano += 1

    crescimento = (pib_cenario / pib_inicial - 1) * 100
    print(f"{nome}: PIB = {pib_cenario:.2f} | crescimento = {crescimento:.2f}%")

    # TODO: atualize o maior PIB e o nome do cenário vencedor
    if pib_cenario > maior_pib:
        maior_pib = pib_cenario
        nome_maior = nome

print(f"Maior PIB: {nome_maior} (R$ {maior_pib:.2f})")
````

### ✍️ Explique (Q8)

Por que `pib_cenario` é reiniciado **dentro** do laço externo? O que aconteceria se ele fosse inicializado uma única vez, antes do `while` externo?

**Sua resposta:**

O PIB do cenário deve ficar dentro do *while* interno, pois é assim que ele obtém os dados de cada cenário e os avalia; se estivesse no *while* externo, ocorreria um erro, já que o sistema não saberia qual taxa avaliar.

## 🏆 Questão 9 — Quem bate a meta primeiro? (1,0)

**Mesa:** Estratégia · **Conceito:** loops aninhados + `break` + flag

A **LEMA Capital** quer a primeira unidade de negócio a atingir **R$ 1.000 milhões** de receita. A busca testa três cenários **nesta ordem**:

- **A:** receita inicial 500, crescimento 3,0% ao ano;
- **B:** receita inicial 620, crescimento 4,5% ao ano;
- **C:** receita inicial 700, crescimento 3,8% ao ano.

O laço externo percorre os cenários; o interno avança **ano a ano, até 15 anos**. Quando a meta é atingida, marque a flag `encontrado`, guarde o cenário e o ano, e use `break` no laço interno. A condição do laço externo deve incluir `not encontrado`.

**Saída esperada:** o cenário **A** não atinge a meta em 15 anos; o cenário **B** atinge no **ano 11**; a busca para e o cenário **C** nem é testado.

> 🧠 **Pergunta crítica:** o cenário **C** atingiria a meta no **ano 10**. A ordem de teste fez a busca devolver **B** — que não é o mais rápido. O que isso revela sobre usar `break` + flag para “primeiro que encontra”?

````python
meta = 1000.0
cenario = 0
encontrado = False
cenario_vencedor = ""
ano_vencedor = 0

while cenario < 3 and not encontrado:
    cenario += 1

    # TODO: defina nome, receita inicial e taxa de cada cenário
    if cenario == 1:
        nome = "A"
        receita = 500.0
        taxa = 0.030
    elif cenario == 2:
        nome = "B"
        receita = 620.0
        taxa = 0.045
    else:
        nome = "C"
        receita = 700.0
        taxa = 0.038

    ano = 0

    # TODO: laço interno de até 15 anos
    while ano < 15:
        ano += 1

        # TODO: atualize a receita do ano
        receita = receita * (1 + taxa)

        # TODO: se a receita >= meta, marque a flag, guarde os dados e use break
        if receita >= meta:
            encontrado = True
            cenario_vencedor = nome
            ano_vencedor = ano
            break

    if not encontrado:
        print(f"Cenário {nome}: meta não atingida em 15 anos (R$ {receita:.2f}).")

if encontrado:
    print(f"Busca encerrada: cenário {cenario_vencedor}, ano {ano_vencedor}.")
else:
    print("Nenhum cenário atingiu a meta.")
````

### ✍️ Explique (Q9)

Por que o `break` sozinho não encerra o laço externo? Para que serve a flag encontrado? E como você faria se quisesse o cenário que atinge a meta **no menor tempo**, e não o primeiro na ordem testada?

**Sua resposta:**
A flag encontrada serve para que o programa pare de executar os outros cenários, já que um deles já teria chegado à meta. Para encontrar o cenário que atinge a meta no menor tempo em vez do primeiro, deve-se eliminar a flag de "encontrado" e remover o break, para que também sejam avaliados os outros cenários.


## 🏆 Questão 10 — O equilíbrio de mercado (1,0)

**Mesa:** Microeconomia · **Conceito:** `while` de convergência + contador de segurança + `break`

Um mercado tem:

- demanda: `Qd = 120 - 2P`
- oferta: `Qs = 20 + 3P`

O preço começa em **R$ 10,00**. A cada iteração, o leiloeiro ajusta o preço na direção do equilíbrio:

`preco = preco + 0.1 * (Qd - Qs)`

O loop para quando `|Qd - Qs| < 0.01` ou quando atingir **100 iterações** (contador de segurança). Ao final, informe:

- se o mercado começou com excesso de demanda ou de oferta;
- o preço de equilíbrio aproximado, `Qd` e `Qs`;
- em quantas iterações convergiu (ou se estourou o limite).

**Saída esperada:** excesso de **demanda** de 50 unidades no preço inicial; convergência em **13 iterações**; preço **R$ 20,00**, `Qd` e `Qs` aproximadamente **80,00**.

**Dica:** para calcular a distância sem `abs()`, copie o excesso para outra variável e, se ela for negativa, troque o sinal.

````python
preco = 10.0
tolerancia = 0.01
limite = 100
iteracao = 0

# TODO: calcule o excesso inicial e classifique o mercado
excesso_inicial = 120 - 2 * preco - (20 + 3 * preco)
if excesso_inicial > 0:
    print(f"O mercado começa com excesso de demanda de {excesso_inicial:.2f}")
elif excesso_inicial < 0:
    print("O mercado começa com excesso de oferta.")
else:
    print("O mercado já está em equilíbrio.")

while iteracao < limite:
    qd = 120 - 2 * preco
    qs = 20 + 3 * preco
    excesso = qd - qs

    # TODO: calcule a distância (valor absoluto) sem usar abs()
    distancia = qd - qs
    if distancia < 0:
        distancia = -distancia

    # TODO: se a distância ficar abaixo da tolerância, use break
    if distancia < tolerancia:
        break

    # TODO: ajuste o preço e incremente a iteração
    preco = preco + excesso * 0.1
    iteracao += 1

# TODO: imprima o resultado, distinguindo convergência de limite atingido
if iteracao >= limite:
    print("⚠️ Limite de iterações atingido sem convergência.")
else:
    print(f"Convergência alcançada em {iteracao} iterações.")
    print(f"Preço de equilíbrio: R$ {preco:.2f}")
    print(f"Quantidade de equilíbrio: {qd:.2f}")
````

### ✍️ Explique (Q10)

Por que o loop termina pelo `break`, e não pela condição `iteracao < limite`? O que isso diz sobre a convergência do preço? O que aconteceria se o ajuste fosse `preco = preco - 0.1 * (Qd - Qs)`?

**Sua resposta:**

El bucle termina mediante un break porque el algoritmo encontró el precio de equilibrio del mercado deseada; por lo tanto, no es necesario alcanzar el número máximo de iteraciones. Se o ajuste fosse `preco = preco - 0.1 * (Qd - Qs), Qd - Qs sempre vai aumentar em vez de diminuir, a condição do break nunca será atingida, el algoritmo en solitario finaliza con el número máximo de interacciones.
